Perguntei para o ChatGPT se existia alguma forma de, usando o python, clicar no texto do print e ele mudar alguma variavel. Ele me entregou uma outra opção criando esses botões utilizando a biblioteca ipywidgets.

```python
import ipywidgets as widgets
from IPython.display import display

variavel = 0

ligar = widgets.Button(description="LIGAR")
desligar = widgets.Button(description="DESLIGAR")

def liga(b):
    global variavel
    variavel = 1
    print("Estado:", variavel)

def desliga(b):
    global variavel
    variavel = 0
    print("Estado:", variavel)

ligar.on_click(liga)
desligar.on_click(desliga)

display(ligar, desligar)
```
